```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
np.random.seed(42)
m = 500
cpu = np.random.uniform(10,95,m)
requests = np.random.uniform(5,500,m)
ram = np.random.uniform(20,90,m)
latency = np.random.uniform(1,50,m)

noise = np.random.normal(0,15,m)
y = 2.1*cpu + 0.08*requests + 1.4*ram + 3.5*latency + 20 + noise
x = np.column_stack([cpu, requests, ram, latency])
```

```python
df = pd.DataFrame(x, columns=["cpu", "requests", "ram", "latency"])
df["response_time"] = y

print(df.head())
print(df.describe())
print(df.isna().sum())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(X_train.shape, X_test.shape)
```

```python
def add_bias(X):
  m = X.shape[0]
  return np.c_[np.ones(m), X]

X_train_b = add_bias(X_train)
X_test_b = add_bias(X_test)

print (X_train_b.shape)
```

```python
def predict(X, theta):
  return X @ theta

def compute_cost(X, y, theta):
  m = len(y)
  errors = predict(X, theta) - y
  return (1/(2*m)) * np.sum(errors**2)

theta0 = np.zeros(X_train_b.shape[1])
print(compute_cost(X_train_b, Y_train, theta0))
```
